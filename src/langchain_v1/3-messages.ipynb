# LangChain Messages

> **Key idea:** Messages are the structured units exchanged between the user, the model, and tools during a LangChain conversation.

## Why messages are used

A message contains a role and content. Keeping the role explicit helps the model understand whether the text is an instruction, a user request, a model response, or a tool result.

## Main message types

### 1. `SystemMessage`

Provides instructions and behavior guidelines for the model. It usually defines the model's role, rules, tone, or constraints.

```python
from langchain_core.messages import SystemMessage

message = SystemMessage(
    content="You are a helpful assistant that explains concepts clearly."
)
```

### 2. `HumanMessage`

Represents the user's input or question.

```python
from langchain_core.messages import HumanMessage

message = HumanMessage(content="Explain LangChain messages.")
```

### 3. `AIMessage`

Represents a response generated by the language model. It may contain normal text, tool calls, reasoning metadata, or usage metadata.

```python
from langchain_core.messages import AIMessage

message = AIMessage(content="Messages represent conversation turns.")
```

### 4. `ToolMessage`

Contains the result returned by a tool after the model requests that tool to run. It is linked to the model's tool call by a `tool_call_id`.

```python
from langchain_core.messages import ToolMessage

message = ToolMessage(
    content="The weather is sunny.",
    tool_call_id="call_123",
)
```

## Typical message flow

A tool-using conversation commonly follows this sequence:

1. `SystemMessage` gives the model its instructions.
2. `HumanMessage` contains the user's request.
3. `AIMessage` requests a tool call.
4. The tool runs with the requested arguments.
5. `ToolMessage` returns the tool result.
6. A final `AIMessage` explains the result to the user.

```text
SystemMessage -> HumanMessage -> AIMessage(tool call)
                                      |
                                      v
                               ToolMessage
                                      |
                                      v
                              AIMessage(answer)
```

## Sending messages to a model

A list of messages can be passed to a chat model:

```python
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="What is LangChain?"),
]

response = model.invoke(messages)
print(response.content)
```

## Important revision points

- `SystemMessage` controls behavior and instructions.
- `HumanMessage` represents user input.
- `AIMessage` represents model output and may include tool calls.
- `ToolMessage` represents the result produced by a tool.
- `content` contains the readable text, while metadata stores additional information.
- Message order matters because the model uses the conversation history as context.


````python
from dotenv import load_dotenv
load_dotenv()
````

````python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model = "groq:openai/gpt-oss-20b",
    temperature = 0.7,
)
````

````python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a poetry expert"),
    HumanMessage(content="Write a poem on my car"),
]

response = model.invoke(messages)
print(response.content)
````

````python
messages.append(response)
messages.append(HumanMessage(content="python coding language"))

response = model.invoke(messages)
print(response.content)
````
